## 1. Project Overview

Fish weight is an important characteristic that can be estimated
from measurable physical characteristics of a fish.

In this project, we will analyze a fish dataset and investigate
the relationship between fish weight and characteristics such as
length, height, and width.

After performing data cleaning and exploratory data analysis,
a Linear Regression model will be developed to predict fish weight.

The performance of the model will be evaluated using appropriate
regression evaluation metrics.

## 2. Problem Statement

The objective is to develop a machine learning model that can
predict the weight of a fish using its physical characteristics.

The model will use features such as:

- Species
- Vertical length (Length1)
- Diagonal length (Length2)
- Cross length (Length3)
- Height
- Width

The predicted weight will then be compared with the actual weight
to evaluate the performance of the Linear Regression model.
