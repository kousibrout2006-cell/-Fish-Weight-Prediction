```python
import pandas as pd

url = "https://gist.githubusercontent.com/5chang2/6923330a794da4f4e05c06599c592914/raw/8ffaa770a44541cded1146b524820ef6c50856ec/fish.csv"

df = pd.read_csv(url)

# Remove duplicate records
df = df.drop_duplicates()

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
numerical_columns = df.select_dtypes(include="number").columns

print("Numerical columns:")
print(numerical_columns.tolist())
```

```python
df[numerical_columns].describe()
```

```python
Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

print("Q1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nIQR:")
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bounds:")
print(lower_bound)

print("\nUpper bounds:")
print(upper_bound)
```

```python
outlier_counts = {}

for column in numerical_columns:
    lower = lower_bound[column]
    upper = upper_bound[column]

    count = ((df[column] < lower) | (df[column] > upper)).sum()

    outlier_counts[column] = count

print("Outlier counts:")
print(outlier_counts)
```

```python
outlier_summary = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower Bound": lower_bound,
    "Upper Bound": upper_bound,
    "Outlier Count": pd.Series(outlier_counts)
})

outlier_summary
```

# Day 10 - Outlier Analysis

## Objective

The objective of Day 10 was to identify potential outliers in
the numerical variables of the Fish dataset.

## Method

The Interquartile Range (IQR) method was used to identify
potential outliers.

The IQR is calculated as:

IQR = Q3 - Q1

The lower and upper limits were calculated using:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries were flagged as potential
outliers.

## Variables Analyzed

The following numerical variables were analyzed:

- Weight
- Length1
- Length2
- Length3
- Height
- Width

## Visualization

Box plots were created to visually examine the distribution
of the numerical variables and identify potential extreme
values.

A separate box plot was also created for the target variable,
Weight.
